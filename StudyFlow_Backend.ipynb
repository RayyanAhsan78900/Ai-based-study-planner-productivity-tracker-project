## 📊 ML Algorithm Explained

### Priority Score Formula:
```
priority = (difficulty × 3) + (urgency × 2) + (1 / hours_per_day)
```
- **difficulty**: 1=Easy, 2=Medium, 3=Hard
- **urgency**: 1-10 based on days until deadline
- **hours**: More hours = slightly lower priority (already well-covered)

### Completion Prediction:
```
prediction = 100 - (difficulty × 10) + (min(hours,4) × 5) - (priority × 0.5)
```
Gives 10-95% likelihood of completing task on time.

### Schedule Distribution:
- Hard subjects get more study days (up to 4)
- Subjects are interleaved for spaced repetition
- Urgent deadlines trigger front-loading of sessions
